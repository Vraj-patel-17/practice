STEMMING

stemming is the process of reducing a word to its word stem that affixes to suffixes or prefixes or to the roots of word known as lemma. Stemming is important for nlp

PORTERstemming

```python
from nltk.stem import PorterStemmer
```

```python
words=["eats","eaten",'ate',"writes","writing","programming","programmer"]
stemming=PorterStemmer()
for word in words:
    print(word+"--->"+stemming.stem(word))
```

```python
stemming.stem("sitting")
```

```python
stemming.stem("going")
```

```python
stemming.stem("went")
```

Regexstemmer

```python
from nltk.stem import RegexpStemmer
```

```python
reg_stemmer=RegexpStemmer('ing$|s$|e$|able$',min=4)
```

```python
reg_stemmer.stem("eating")
```

Snowball stemmer

```python
from nltk.stem import SnowballStemmer
```

```python
snow_stem=SnowballStemmer("english")
for word in words:
    print(word+"--->"+snow_stem.stem(word))
```

```python
stemming.stem("fairly"),stemming.stem("sportingly")
```

```python
snow_stem.stem("fairly"),snow_stem.stem("sportingly")
```
