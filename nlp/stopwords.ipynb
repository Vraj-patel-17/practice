```python
paragraph="""The present is theirs; the future, for which I really worked, is mine.” These words by Nikola Tesla reflect the spirit of innovation, imagination, and relentless dedication. Tesla believed that true progress comes from thinking beyond the limits of the present and daring to imagine what others cannot yet see. His life reminds us that every great invention, discovery, and achievement begins as an idea in the mind of someone willing to dream differently. Today, as technology continues to shape our world, Tesla's vision inspires us to stay curious, creative, and fearless in pursuing the future."""
```

```python
import nltk
from nltk.stem import PorterStemmer
from nltk.corpus import stopwords
```

```python
stopwords.words("english")
```

```python
from nltk.stem import PorterStemmer
stemmer=PorterStemmer()
```

```python
sentences=nltk.sent_tokenize(paragraph)
```

```python
# apply stopwords and filter and then stemmingt
for i in range(len(sentences)):
    words=nltk.word_tokenize(sentences[i])
    words=[stemmer.stem(word) for word in words if word not in set(stopwords.words())]
    sentences[i]=" ".join(words)
```

```python
from nltk.stem import SnowballStemmer
```

```python
snow_stem=SnowballStemmer("english")
```

```python
for i in range(len(sentences)):
    words=nltk.word_tokenize(sentences[i])
    words=[snow_stem.stem(word) for word in words if word not in set(stopwords.words())]
    sentences[i]=" ".join(words)
```

```python
sentences
```

```python
from nltk.stem import WordNetLemmatizer
```

```python
lemmatizer=WordNetLemmatizer()
```

```python
for i in range(len(sentences)):
    words=nltk.word_tokenize(sentences[i])
    words=[lemmatizer.lemmatize(word) for word in words if word not in set(stopwords.words())]
    sentences[i]=" ".join(words)
```
