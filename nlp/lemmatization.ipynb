Lemmatization is a technique like stemming. The output of lemmatiztion is called lemma , which is root word rather than root stem ,the output of stemming

```python
from nltk.stem import WordNetLemmatizer
```

```python
lemmatizer=WordNetLemmatizer()
```

```python
lemmatizer.lemmatize("going")
```

```python
'''pos 
verb-v
adjective-a
adverb-r
noun-n'''
lemmatizer.lemmatize("going",pos="v")
```

```python
words=["eats","eaten",'ate',"writes","writing","programming","programmer"]
for word in words:
    print(word+"--->"+lemmatizer.lemmatize(word,pos="v"))
```
