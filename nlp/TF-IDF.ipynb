```python
import pandas as pd
```

```python
messages=pd.read_csv('SMSSpamCollection',sep='\t',names=["label","message"])
```

```python
messages
```

```python
import re
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
wordlemmatizer=WordNetLemmatizer()
```

```python
corpus=[]
for i in range(len(messages)):
    review=re.sub('[^a-zA-Z]',' ',messages['message'][i]).lower().split()
    review=[wordlemmatizer.lemmatize(word) for word in review if word not in stopwords.words("english")]
    review=' '.join(review)
    corpus.append(review)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tfidf=TfidfVectorizer(max_features=100)
x=tfidf.fit_transform(corpus).toarray()
```

```python
import numpy as np
np.set_printoptions(edgeitems=30,linewidth=100,formatter=dict(float=lambda x : "%.3g" % x))
```

```python
# ngrams
tfidf=TfidfVectorizer(max_features=100,ngram_range=(2,2))
x=tfidf.fit_transform(corpus).toarray()
```
