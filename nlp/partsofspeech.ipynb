```python
import nltk
nltk.download('averaged_perceptron_tagger_eng')
```

```python
from nltk.corpus import stopwords
```

```python
paragraph="""The present is theirs; the future, for which I really worked, is mine.” These words by Nikola Tesla reflect the spirit of innovation, imagination, and relentless dedication. Tesla believed that true progress comes from thinking beyond the limits of the present and daring to imagine what others cannot yet see. His life reminds us that every great invention, discovery, and achievement begins as an idea in the mind of someone willing to dream differently. Today, as technology continues to shape our world, Tesla's vision inspires us to stay curious, creative, and fearless in pursuing the future."""
```

```python
sentence=nltk.sent_tokenize(paragraph)
```

```python
for i in range(len(sentence)):
    words=nltk.word_tokenize(sentence[i])
    words=[word for word in words if word not in set(stopwords.words("english"))]
    pos_tag=nltk.pos_tag(words)
    print(pos_tag)
```
