```python
import pandas as pd
messages=pd.read_csv('SMSSpamCollection',sep='\t',names=["label","message"])
```

```python
import re
from nltk.stem import PorterStemmer
from nltk.corpus import stopwords
ps=PorterStemmer()
```

```python
corpus=[]
for i in range(len(messages)):
    review=re.sub('[^a-zA-Z]',' ',messages['message'][i]).lower().split()
    review=[ps.stem(word) for word in review if word not in stopwords.words("english")]
    review=' '.join(review)
    corpus.append(review)
```

```python
y=pd.get_dummies(messages['label'])
y=y.iloc[:,0].values
```

```python
from sklearn.model_selection import train_test_split
xtr,xts,ytr,yts=train_test_split(corpus,y,test_size=0.2)
```

USING BAG OF WORDS

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
cv=CountVectorizer(max_features=2500,ngram_range=(1,2))
```

```python
xtr=cv.fit_transform(xtr).toarray()
xts=cv.transform(xts).toarray()
```

```python
from sklearn.naive_bayes import MultinomialNB
```

```python
spam_detect_model=MultinomialNB().fit(xtr,ytr)
```

```python
ypred=spam_detect_model.predict(xts)
```

```python
from sklearn.metrics import classification_report,accuracy_score
```

```python
accuracy_score(yts,ypred)
```

```python
print(classification_report(yts,ypred))
```
