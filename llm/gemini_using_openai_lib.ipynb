```python
import dotenv
import os
```

```python
dotenv.load_dotenv()
```

```python
gemini_api_key=os.getenv('GEMINI_API_KEY')
if gemini_api_key:
    print("API key loaded successfully")
else:
    print("Failed to load API key")
```

#checking models supported by api

```python
import google.generativeai as genai

#configure api key
genai.configure(api_key=gemini_api_key)
for model in genai.list_models():
    if 'generateContent' in model.supported_generation_methods:
        print(f"Model Name: {model.name}")
        print(f"Display Name: {model.display_name}")
        print(f"Input Token Limit: {model.input_token_limit}\n")
```

```python
gemini=OpenAI(api_key=gemini_api_key,base_url="https://generativelanguage.googleapis.com/v1beta/openai/")
```

```python
response=gemini.chat.completions.create(model="gemini-2.5-flash",messages=[{"role":"user","content":"Give me a scientific fun fact"}])
print(response.choices[0].message.content)
```
