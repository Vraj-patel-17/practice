```python
import tensorflow as tf
from tensorflow import keras
```

```python
fashion_data=keras.datasets.fashion_mnist
(x_train,y_train),(x_test,y_test)=fashion_data.load_data()
```

```python
x_train.shape,x_train.dtype
```

```python
#creating validation set
#dividing by 255 to scale pixel intensites
x_valid,x_tr=x_train[:5000]/255.0,x_train[5000:]/255.0
y_valid,y_tr=y_train[:5000],y_train[5000:]
```

```python
class_names = ["T-shirt/top", "Trouser", "Pullover", "Dress", "Coat",
"Sandal", "Shirt", "Sneaker", "Bag", "Ankle boot"]
```

```python
#creating model
model=keras.models.Sequential()
model.add(keras.layers.Flatten(input_shape=[28,28]))
model.add(keras.layers.Dense(300,activation='relu'))
model.add(keras.layers.Dense(100,activation='relu'))
model.add(keras.layers.Dense(10,activation='softmax')) 
```

```python
model.summary()
```

```python
weights,biases=model.layers[1].get_weights()
weights,weights.shape
```

```python
biases,biases.shape
```

```python
model.compile(loss='sparse_categorical_crossentropy',optimizer='sgd',metrics=["accuracy"])
```

```python
history=model.fit(x_tr,y_tr,validation_data=(x_valid,y_valid),epochs=30)
```

```python
history.params
```

```python
history.epoch,history.history
```

```python
model.evaluate(x_test,y_test)
```
